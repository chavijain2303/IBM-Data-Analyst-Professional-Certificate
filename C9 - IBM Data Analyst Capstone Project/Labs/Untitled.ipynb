```python
# If wget does not work on Windows/Jupyter,
# directly load the dataset from the URL

import pandas as pd
import matplotlib.pyplot as plt

# Dataset URL
file_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"

# Load dataset
df = pd.read_csv(file_url)

# Display first few rows
df.head()
```

```python
# Count respondents by age group

age_counts = (
    df['Age']
    .dropna()
    .value_counts()
)

# Arrange age groups logically

age_order = [
    'Under 18 years old',
    '18-24 years old',
    '25-34 years old',
    '35-44 years old',
    '45-54 years old',
    '55-64 years old',
    '65 years or older'
]

age_counts = age_counts.reindex(age_order)

plt.figure(figsize=(10,6))

plt.plot(
    age_counts.index,
    age_counts.values,
    marker='o',
    linewidth=3
)

plt.title('Age Distribution of Respondents')
plt.xlabel('Age Group')
plt.ylabel('Number of Respondents')

plt.xticks(rotation=45)

plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

mainbranch_counts = (
    df['MainBranch']
    .dropna()
    .value_counts()
)

fig, ax = plt.subplots(figsize=(10,8))

wedges, texts, autotexts = ax.pie(
    mainbranch_counts.values,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.8,
    wedgeprops={'width':0.4}
)

ax.set_title(
    'Professional Background of Respondents',
    fontsize=14,
    fontweight='bold'
)

# Put labels in legend
ax.legend(
    wedges,
    mainbranch_counts.index,
    title='MainBranch',
    loc='center left',
    bbox_to_anchor=(1, 0.5)
)

plt.tight_layout()
plt.show()
```

```python
# Top 10 countries

country_counts = (
    df['Country']
    .dropna()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12,6))

plt.bar(
    country_counts.index,
    country_counts.values
)

plt.title('Top 10 Countries by Respondent Count')
plt.xlabel('Country')
plt.ylabel('Number of Respondents')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.read_csv('popular-languages.csv')

print(df.columns)
print(df.head())
```

```python
# Convert salary column to numeric
df['Average Annual Salary'] = (
    df['Average Annual Salary']
    .replace('[\$,]', '', regex=True)
    .astype(float)
)

# Sort by salary (descending)
df_sorted = df.sort_values(
    by='Average Annual Salary',
    ascending=False
)

# Create bar chart
plt.figure(figsize=(10,6))

plt.bar(
    df_sorted['Language'],
    df_sorted['Average Annual Salary']
)

plt.title('Average Annual Salary by Programming Language')
plt.xlabel('Programming Language')
plt.ylabel('Average Annual Salary (USD)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# If wget does not work on Windows/Jupyter,
# directly load the dataset from the URL

import pandas as pd
import matplotlib.pyplot as plt

# Dataset URL
file_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"

# Load dataset
df = pd.read_csv(file_url)

# Display first few rows
df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Extract languages currently used
current_lang = (
    df['LanguageHaveWorkedWith']
    .dropna()
    .str.split(';')
    .explode()
)

# Top 10 languages
current_counts = current_lang.value_counts().head(10)

# Plot
plt.figure(figsize=(10,6))

current_counts.sort_values().plot(
    kind='barh'
)

plt.title('Top 10 Programming Languages Used (Current Year)')
plt.xlabel('Number of Respondents')
plt.ylabel('Programming Language')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ==================================================
# DEMOGRAPHICS DASHBOARD
# ==================================================

fig, axes = plt.subplots(2, 2, figsize=(18, 12))

fig.suptitle(
    'Demographics Dashboard',
    fontsize=20,
    fontweight='bold'
)

# ==================================================
# 1. TOP COUNTRIES
# ==================================================

country_counts = (
    df['Country']
    .dropna()
    .value_counts()
    .head(10)
)

country_counts.sort_values().plot(
    kind='barh',
    ax=axes[0, 0]
)

axes[0, 0].set_title('Top 10 Countries')
axes[0, 0].set_xlabel('Number of Respondents')
axes[0, 0].set_ylabel('Country')

# ==================================================
# 2. AGE DISTRIBUTION (PIE CHART)
# ==================================================

age_counts = (
    df['Age']
    .dropna()
    .value_counts()
)

top_age = age_counts.head(5)
other_age = age_counts.iloc[5:].sum()

age_dashboard = pd.concat([
    top_age,
    pd.Series({'Others': other_age})
])

wedges, texts, autotexts = axes[0, 1].pie(
    age_dashboard.values,
    autopct='%1.1f%%',
    startangle=90
)

axes[0, 1].set_title('Age Distribution')

axes[0, 1].legend(
    wedges,
    age_dashboard.index,
    title='Age Groups',
    loc='upper center',
    bbox_to_anchor=(0.5, -0.10),
    ncol=2,
    fontsize=8
)

# ==================================================
# 3. EDUCATION LEVEL
# ==================================================

education_counts = (
    df['EdLevel']
    .dropna()
    .value_counts()
    .head(10)
)

education_counts.sort_values().plot(
    kind='barh',
    ax=axes[1, 0]
)

axes[1, 0].set_title('Education Level')
axes[1, 0].set_xlabel('Number of Respondents')
axes[1, 0].set_ylabel('Education')

# ==================================================
# 4. PROFESSIONAL BACKGROUND (DONUT CHART)
# ==================================================

mainbranch_counts = (
    df['MainBranch']
    .dropna()
    .value_counts()
)

top_branch = mainbranch_counts.head(5)
other_branch = mainbranch_counts.iloc[5:].sum()

branch_dashboard = pd.concat([
    top_branch,
    pd.Series({'Others': other_branch})
])

wedges2, texts2, autotexts2 = axes[1, 1].pie(
    branch_dashboard.values,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4}
)

axes[1, 1].set_title('Professional Background')

axes[1, 1].legend(
    wedges2,
    branch_dashboard.index,
    title='MainBranch',
    loc='upper center',
    bbox_to_anchor=(0.5, -0.15),
    ncol=2,
    fontsize=8
)

# ==================================================
# SPACING
# ==================================================

plt.subplots_adjust(
    top=0.90,
    hspace=0.60,
    wspace=0.40
)

plt.show()
```
