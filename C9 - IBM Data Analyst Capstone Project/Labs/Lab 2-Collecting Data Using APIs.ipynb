 ##### The keys in the json are

You can also view  the json file contents  from the following URL.

https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json

```python
# API URL
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

# Function to get number of jobs for a technology
def get_number_of_jobs_T(technology):
    
    # Retrieve the full dataset from API
    response = requests.get(api_url)
    
    # Convert JSON data into Python object
    data = response.json()
    
    # Convert to DataFrame
    df = pd.DataFrame(data)
    
    # Filter rows where technology exists in 'Key Skills'
    number_of_jobs = df[df['Key Skills'].str.contains(technology, case=False, na=False)].shape[0]
    
    return technology, number_of_jobs
```

Calling the function for Python and checking if it works.

```python
get_number_of_jobs_T("Python")
```

#### Write a function to find number of jobs in US for a location of your choice

```python
def get_number_of_jobs_L(location):
    
    # Retrieve the full dataset from API
    response = requests.get(api_url)
    
    # Convert JSON data into Python object
    data = response.json()
    
    # Convert to DataFrame
    df = pd.DataFrame(data)
    
    # Filter rows where location exists in 'Location'
    number_of_jobs = df[df['Location'].str.contains(location, case=False, na=False)].shape[0]
    
    return location, number_of_jobs
```
