### Task 1: Exploring Relationships with Scatter Plots

#### 1. Scatter Plot for Age vs. Job Satisfaction

Visualize the relationship between respondents' age (`Age`) and job satisfaction (`JobSatPoints_6`). Use this plot to identify any patterns or trends.

```python
# Display first few rows of required columns

print(df[['Age', 'JobSatPoints_6']].head())
```

```python
# Convert Age categories into approximate numeric values

age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29,
    "35-44 years old": 39,
    "45-54 years old": 49,
    "55-64 years old": 59,
    "65 years or older": 70
}

df['Age_numeric'] = df['Age'].map(age_mapping)
```

```python
# Convert JobSatPoints_6 to numeric

df['JobSatPoints_6'] = pd.to_numeric(
    df['JobSatPoints_6'],
    errors='coerce'
)
```

```python
# Remove missing values

scatter_df = df[
    ['Age_numeric', 'JobSatPoints_6']
].dropna()
```

```python
# Create scatter plot

plt.figure(figsize=(10,6))

plt.scatter(
    scatter_df['Age_numeric'],
    scatter_df['JobSatPoints_6'],
    alpha=0.5
)

plt.title('Age vs Job Satisfaction')
plt.xlabel('Age')
plt.ylabel('Job Satisfaction')

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

### Task 3: Combining Scatter Plots with Additional Features

#### 1. Bubble Plot of Compensation vs. Job Satisfaction with Age as Bubble Size

Create a bubble plot to explore the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`), with bubble size representing age.

```python
# Convert required columns to numeric

df['ConvertedCompYearly'] = pd.to_numeric(
    df['ConvertedCompYearly'],
    errors='coerce'
)

df['JobSatPoints_6'] = pd.to_numeric(
    df['JobSatPoints_6'],
    errors='coerce'
)
```

```python
# Convert Age categories into approximate numeric values

age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29,
    "35-44 years old": 39,
    "45-54 years old": 49,
    "55-64 years old": 59,
    "65 years or older": 70
}

df['Age_numeric'] = df['Age'].map(age_mapping)
```

```python
# Remove missing values

bubble_df = df[
    ['ConvertedCompYearly', 'JobSatPoints_6', 'Age_numeric']
].dropna()
```

```python
# Remove extreme compensation outliers for better visualization

upper_limit = bubble_df['ConvertedCompYearly'].quantile(0.99)

bubble_df = bubble_df[
    bubble_df['ConvertedCompYearly'] <= upper_limit
]
```

```python
# Create bubble plot

plt.figure(figsize=(12,7))

plt.scatter(
    bubble_df['ConvertedCompYearly'],
    bubble_df['JobSatPoints_6'],
    s=bubble_df['Age_numeric'] * 3,   # Bubble size based on age
    alpha=0.5
)

plt.title('Compensation vs Job Satisfaction (Bubble Size = Age)')
plt.xlabel('ConvertedCompYearly')
plt.ylabel('Job Satisfaction')

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

#### 2. Scatter Plot for Popular Programming Languages by Job Satisfaction

Visualize the popularity of programming languages (`LanguageHaveWorkedWith`) against job satisfaction using a scatter plot. Use points to represent satisfaction levels for each language.

```python
# Select required columns

lang_df = df[
    ['LanguageHaveWorkedWith', 'JobSatPoints_6']
].dropna()

# Display first few rows
print(lang_df.head())
```

```python
# Convert JobSatPoints_6 to numeric

lang_df['JobSatPoints_6'] = pd.to_numeric(
    lang_df['JobSatPoints_6'],
    errors='coerce'
)

# Remove missing values
lang_df = lang_df.dropna()
```

```python
# Split multiple languages into separate rows

lang_exploded = lang_df.assign(
    LanguageHaveWorkedWith=
    lang_df['LanguageHaveWorkedWith'].str.split(';')
).explode('LanguageHaveWorkedWith')
```

```python
# Find top 10 most popular languages

top_languages = (
    lang_exploded['LanguageHaveWorkedWith']
    .value_counts()
    .head(10)
    .index
)

print(top_languages)
```

```python
# Filter only top languages

filtered_df = lang_exploded[
    lang_exploded['LanguageHaveWorkedWith'].isin(top_languages)
].copy()
```

```python
# Calculate average job satisfaction for each language

language_satisfaction = (
    filtered_df
    .groupby('LanguageHaveWorkedWith')['JobSatPoints_6']
    .mean()
    .reset_index()
)

print(language_satisfaction)
```

```python
# Create scatter plot

plt.figure(figsize=(12,6))

plt.scatter(
    language_satisfaction['LanguageHaveWorkedWith'],
    language_satisfaction['JobSatPoints_6'],
    s=150,
    alpha=0.7
)

plt.title('Programming Languages vs Job Satisfaction')
plt.xlabel('Programming Language')
plt.ylabel('Average Job Satisfaction')

plt.xticks(rotation=30, ha='right')

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Remove extreme compensation outliers

upper_limit = employment_df['ConvertedCompYearly'].quantile(0.99)

employment_df = employment_df[
    employment_df['ConvertedCompYearly'] <= upper_limit
]
```

```python
# Create scatter plot with employment type categories

plt.figure(figsize=(12,7))

sns.scatterplot(
    data=employment_df,
    x='ConvertedCompYearly',
    y='JobSatPoints_6',
    hue='Employment',
    alpha=0.6
)

plt.title('Compensation vs Job Satisfaction by Employment Type')
plt.xlabel('ConvertedCompYearly')
plt.ylabel('Job Satisfaction')

plt.grid(True, linestyle='--', alpha=0.5)

plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

#### 2. Scatter Plot for Work Experience vs. Age Group by Country

Compare work experience (`YearsCodePro`) across different age groups (`Age`) and countries (`Country`). Use colors to represent different countries and markers for age groups.

```python
# Select required columns

country_df = df[
    ['Age', 'YearsCodePro', 'Country']
].dropna()

# Display first few rows
print(country_df.head())
```

```python
# Convert YearsCodePro to numeric

country_df['YearsCodePro'] = pd.to_numeric(
    country_df['YearsCodePro'],
    errors='coerce'
)
```

```python
# Convert Age categories into approximate numeric values

age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29,
    "35-44 years old": 39,
    "45-54 years old": 49,
    "55-64 years old": 59,
    "65 years or older": 70
}

country_df['Age_numeric'] = (
    country_df['Age']
    .map(age_mapping)
)
```
