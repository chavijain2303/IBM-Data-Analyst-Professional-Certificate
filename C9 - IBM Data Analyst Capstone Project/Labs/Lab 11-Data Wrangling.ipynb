<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>

```python
#🔹 Step 1: Create a mapping dictionary for Country
#Based on your dataset inconsistencies:

country_map = {
    "United States of America": "United States",
    "United Kingdom of Great Britain and Northern Ireland": "United Kingdom",
    "Iran, Islamic Republic of...": "Iran",
    "Venezuela, Bolivarian Republic of...": "Venezuela",
    "Republic of Korea": "South Korea",
    "Republic of Moldova": "Moldova",
    "Democratic People's Republic of Korea": "North Korea",
    "Congo, Republic of the...": "Congo",
    "Democratic Republic of the Congo": "Congo (DRC)",
    "Syrian Arab Republic": "Syria",
    "Lao People's Democratic Republic": "Laos"
}
```

```python
#Apply mapping to standardize Country column
df['Country'] = df['Country'].replace(country_map)
```

```python
#Handle whitespace and casing issues (important)
df['Country'] = df['Country'].str.strip()
```

```python
df['Country'] = df['Country'].str.title()
```

```python
#Standardizing EdLevel (Education Level)
#🔹 Step 4: Create mapping for EdLevel
#Example (based on Stack Overflow survey style data):

edlevel_map = {
    "Bachelor’s degree (B.Sc., B.A., etc.)": "Bachelor's",
    "Master’s degree (M.Sc., M.A., etc.)": "Master's",
    "Professional degree (JD, MD, etc.)": "Professional",
    "Some college/university study without earning a degree": "Some College",
    "Secondary school (e.g. American high school, German Realschule)": "High School",
    "Primary/elementary school": "Primary School",
    "Other doctoral degree (Ph.D., Ed.D., etc.)": "PhD"
}
```

```python
#Apply EdLevel mapping
df['EdLevel'] = df['EdLevel'].replace(edlevel_map)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>

```python
#Check missing values in the column
df['ConvertedCompYearly'].isnull().sum()
```

```python
#Impute using MEAN (common approach)
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(df['ConvertedCompYearly'].mean())
```

```python
#Impute using MEDIAN (more robust to outliers ⭐ recommended)
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(df['ConvertedCompYearly'].median())
```

```python
#Verify missing values are handled
df['ConvertedCompYearly'].isnull().sum()
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>

```python
#Check missing values
df['RemoteWork'].isnull().sum()
```

```python
#Find the most frequent value (mode)
df['RemoteWork'].mode()[0]
```

```python
#Fill missing values with mode
df['RemoteWork'] = df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])
```

```python
#Verify missing values are handled
df['RemoteWork'].isnull().sum()
```

### 7. Feature Engineering

<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>

```python
#Check data type (important)
df['YearsCodePro'].unique()
```

```python
#Clean the column (convert to numeric safely)
df['YearsCodePro'] = df['YearsCodePro'].replace({
    'Less than 1 year': 0,
    'More than 50 years': 50
})
```

```python
#Convert to numeric:

df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
```

```python
#Create ExperienceLevel column
def experience_level(years):
    if years < 2:
        return "Beginner"
    elif years < 5:
        return "Intermediate"
    elif years < 10:
        return "Advanced"
    else:
        return "Expert"

df['ExperienceLevel'] = df['YearsCodePro'].apply(experience_level)
```

```python
#Verify result
df[['YearsCodePro', 'ExperienceLevel']].head()
```
