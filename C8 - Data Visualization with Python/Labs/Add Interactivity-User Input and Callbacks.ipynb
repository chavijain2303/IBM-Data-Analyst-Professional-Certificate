# TASK 3 — Add Callback Decorator

Dash callbacks:
- Take input from components
- Process data
- Update output automatically

## Callback Structure

```python
@app.callback(
    Output(...),
    Input(...)
)
```

In this lab:
- Input → Year entered by user
- Output → Updated line chart

````python
@app.callback(
    Output(
        component_id='line-plot',
        component_property='figure'
    ),
    
    Input(
        component_id='input-year',
        component_property='value'
    )
)

def update_graph(selected_year):
    
    # sample dataframe filtering
    filtered_data = airline_data[airline_data['Year'] == selected_year]

    fig = px.line(
        filtered_data,
        x='Month',
        y='ArrDelay'
    )

    return fig
````

# TASK 3 — Create Callback Function

The callback function:
1. Filters data by year
2. Groups data by month
3. Computes average arrival delay
4. Creates line chart
5. Returns updated graph

````python
# Create callback function

@app.callback(
    
    Output(
        component_id='line-plot',
        component_property='figure'
    ),
    
    Input(
        component_id='input-year',
        component_property='value'
    )
)

def get_graph(entered_year):
    
    # Filter data by entered year
    df = airline_data[
        airline_data['Year'] == int(entered_year)
    ]
    
    # Group data by month
    line_data = df.groupby('Month')['ArrDelay'].mean().reset_index()
    
    # Create line chart
    fig = go.Figure(
        
        data=go.Scatter(
            x=line_data['Month'],
            y=line_data['ArrDelay'],
            
            mode='lines',
            
            marker=dict(color='green')
        )
    )
    
    # Update layout
    fig.update_layout(
        title='Month vs Average Flight Delay Time',
        xaxis_title='Month',
        yaxis_title='ArrDelay'
    )
    
    return fig
````

# TASK 4 — Create Dashboard Layout

````python
# Create layout for bar chart dashboard

bar_app.layout = html.Div(children=[
    
    html.H1(
        'Total number of flights to the destination state split by reporting airline',
        
        style={
            'textAlign': 'center',
            'color': '#503D36',
            'font-size': 40
        }
    ),
    
    html.Div(
        [
            "Input Year: ",
            
            dcc.Input(
                id='input-year-bar',
                value='2010',
                type='number',
                
                style={
                    'height': '50px',
                    'font-size': 35
                }
            )
        ],
        
        style={'font-size': 40}
    ),
    
    html.Br(),
    html.Br(),
    
    html.Div(
        dcc.Graph(id='bar-plot')
    )
    
])
````

# TASK 4 — Add Callback Function for Bar Chart

The callback:
- Filters data by year
- Groups flights by destination state
- Creates bar chart
- Updates graph dynamically

````python
# Callback for bar chart dashboard

@bar_app.callback(
    
    Output(
        component_id='bar-plot',
        component_property='figure'
    ),
    
    Input(
        component_id='input-year-bar',
        component_property='value'
    )
)

def get_bar_chart(entered_year):
    
    # Filter data
    df = airline_data[
        airline_data['Year'] == int(entered_year)
    ]
    
    # Group by destination state
    bar_data = df.groupby('DestState')['Flights'].sum().reset_index()
    
    # Create bar chart
    fig = px.bar(
        bar_data,
        
        x='DestState',
        y='Flights',
        
        title='Total number of flights to the destination state split by reporting airline'
    )
    
    # Update layout
    fig.update_layout(
        title='Flights to Destination State',
        xaxis_title='DestState',
        yaxis_title='Flights'
    )
    
    return fig
````
