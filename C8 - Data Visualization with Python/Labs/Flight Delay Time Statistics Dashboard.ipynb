# TASK 4 — Supporting Function

The helper function:
- Filters data by selected year
- Computes average delays
- Returns processed dataframes

Delay types:
- Carrier Delay
- Weather Delay
- NAS Delay
- Security Delay
- Late Aircraft Delay

````python
# Supporting function

def compute_info(airline_data, entered_year):
    
    # Filter data by year
    df = airline_data[
        airline_data['Year'] == int(entered_year)
    ]
    
    # Average carrier delay
    avg_car = df.groupby(
        ['Month', 'Reporting_Airline']
    )['CarrierDelay'].mean().reset_index()
    
    # Average weather delay
    avg_weather = df.groupby(
        ['Month', 'Reporting_Airline']
    )['WeatherDelay'].mean().reset_index()
    
    # Average NAS delay
    avg_NAS = df.groupby(
        ['Month', 'Reporting_Airline']
    )['NASDelay'].mean().reset_index()
    
    # Average security delay
    avg_sec = df.groupby(
        ['Month', 'Reporting_Airline']
    )['SecurityDelay'].mean().reset_index()
    
    # Average late aircraft delay
    avg_late = df.groupby(
        ['Month', 'Reporting_Airline']
    )['LateAircraftDelay'].mean().reset_index()
    
    return avg_car, avg_weather, avg_NAS, avg_sec, avg_late
````

# TASK 5 — Add Callback Decorator

The callback:
- Takes year as input
- Updates five graphs simultaneously

Outputs:
1. carrier-plot
2. weather-plot
3. nas-plot
4. security-plot
5. late-plot

````python
@app.callback(
    [
        Output('carrier-plot', 'figure'),
        Output('weather-plot', 'figure'),
        Output('nas-plot', 'figure'),
        Output('security-plot', 'figure'),
        Output('late-plot', 'figure')
    ],
    
    Input('input-year', 'value')
)

def update_graphs(selected_year):

    # filter data
    df_year = airline_data[airline_data['Year'] == selected_year]

    # create figures
    carrier_fig = px.line(df_year, x='Month', y='CarrierDelay', title='Carrier Delay')
    
    weather_fig = px.line(df_year, x='Month', y='WeatherDelay', title='Weather Delay')
    
    nas_fig = px.line(df_year, x='Month', y='NASDelay', title='NAS Delay')
    
    security_fig = px.line(df_year, x='Month', y='SecurityDelay', title='Security Delay')
    
    late_fig = px.line(df_year, x='Month', y='LateAircraftDelay', title='Late Aircraft Delay')

    return carrier_fig, weather_fig, nas_fig, security_fig, late_fig
````

# TASK 5 — Create Callback Function

The callback function:
1. Computes delay information
2. Creates five line charts
3. Returns updated figures

````python
# Callback function

@app.callback(
    
    [
        Output('carrier-plot', 'figure'),
        Output('weather-plot', 'figure'),
        Output('nas-plot', 'figure'),
        Output('security-plot', 'figure'),
        Output('late-plot', 'figure')
    ],
    
    Input('input-year', 'value')
)

def get_graph(entered_year):
    
    # Compute information
    avg_car, avg_weather, avg_NAS, avg_sec, avg_late = compute_info(
        airline_data,
        entered_year
    )
    
    # Carrier delay graph
    carrier_fig = px.line(
        avg_car,
        
        x='Month',
        y='CarrierDelay',
        color='Reporting_Airline',
        
        title='Average carrier delay time (minutes) by airline'
    )
    
    # Weather delay graph
    weather_fig = px.line(
        avg_weather,
        
        x='Month',
        y='WeatherDelay',
        color='Reporting_Airline',
        
        title='Average weather delay time (minutes) by airline'
    )
    
    # NAS delay graph
    nas_fig = px.line(
        avg_NAS,
        
        x='Month',
        y='NASDelay',
        color='Reporting_Airline',
        
        title='Average NAS delay time (minutes) by airline'
    )
    
    # Security delay graph
    sec_fig = px.line(
        avg_sec,
        
        x='Month',
        y='SecurityDelay',
        color='Reporting_Airline',
        
        title='Average security delay time (minutes) by airline'
    )
    
    # Late aircraft delay graph
    late_fig = px.line(
        avg_late,
        
        x='Month',
        y='LateAircraftDelay',
        color='Reporting_Airline',
        
        title='Average late aircraft delay time (minutes) by airline'
    )
    
    return [
        carrier_fig,
        weather_fig,
        nas_fig,
        sec_fig,
        late_fig
    ]
````
